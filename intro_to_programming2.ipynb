### 1. Write a function that takes a list of strings and a number K. The function should 
### return a new list containing only the strings that have a length greater than K. 
### Input: list=["apple", "pie", "banana"], K=3 
### Output: ['apple', 'banana']

```python
my_list = []    
def my_function(K ,str_list):
    for i in str_list:
        if len(i)>K:
            my_list.append(i)

    return my_list

list=["apple", "pie", "banana"]
K= 3
result = my_function(K, list)
print(result)
```

### 2. Write a function named area() that calculates the area of either a "circle" or a 
### "rectangle". The function should take the shape name as the first argument, 
### followed by the necessary dimensions (radius for circle; length and width for 
 ### rectangle). 
### Example Input: area("circle", 5) OR area("rectangle", 4, 6) 
### Example Output: Circle Area: 78.5, Rectangle Area: 24 

```python
def area(shape_name, **kwargs):
    if shape_name == "rectangle":
        return kwargs['length'] * kwargs['width']
    elif shape_name == "circle":
        return 3.14 * (kwargs['radius'] ** 2)
    else:
        return "Shape not recognized"

print("Circle Area:", area("circle", radius=5))
print("Rectangle Area:", area("rectangle", length=4, width=6))
```

### 4. Write a program to find the sum of all numbers in a list. ensure 
### the program ignores any elements in the list that are not numbers (like strings or 
### booleans). 
### Example Input: [10, "apple", 20, True, 5] 
### Example Output: Total Sum: 35

```python
input_list = [10, "apple", 20, True, 5]
sum_of_integers = 0
for item in input_list:
    if isinstance(item, int):
        sum_of_integers += item
    else:
        continue
    
print(f"Sum of integers: {sum_of_integers}")
```

### 5. Write a Python class Restaurant with attributes like menu_items, book_table, and 
### customer_orders, and methods like add_item_to_menu, book_tables, and 
### customer_order. (Note: Use dictionaries and lists to store the data.) 
### Perform the following tasks now: 
### Now add items to the menu. 
### Make table reservations. 
### Take customer orders. 
### Print the menu. 
### Print table reservations. 
### Print customer orders. 

```python
class Restaurant:
    def __init__(self, menu_items, book_table,customer_orders):
        self.menu_items = menu_items
        self.book_table = book_table
        self.customer_orders = customer_orders

    def add_item_to_menu(self, item):
        self.menu_items.append(item)
        print(f"{item} is added to  {self.menu_items}.")

    def book_tables(self):
        print(f"table {self.book_table} is now booked!")

    def add_customer_orders(self, order):
        for item in order:
            if item in self.menu_items:
                self.customer_orders.append(item)
                print(f"{item} is available and added to the order.")
            else:
                print(f"{item} is not available in the menu.")

    

restaurant = Restaurant(["Pizza", "Pasta", "Salad"], 5, ["Pizza", "Salad"])
restaurant.add_item_to_menu("Burger")
restaurant.book_tables()
restaurant.add_customer_orders(["Pasta", "Burger", "Sushi"])  
print("Customer Orders:", restaurant.customer_orders, "Menu Items:", restaurant.menu_items, "Booked Tables:", restaurant.book_table)

```

#### 6- A Bike Rental System Let us start writing the class Customer that is composed of the following 
#### attributes: 
####  name 
####  age 
#### Here are some of the methods that you need to add to your class Customer; 
####  def __init__(self). 
####  Getter / setter for all attributes. 
#### class MainBikeRental that is composed of the following attributes: 
####  List of bikes 
####  List of Rentals 
#### Here are some of the methods that you need to add to your class MainBikeRental; 
####  def __init__(self). 
####  Getter / setter for all attributes. 
####  requestBike: Takes a request from the customer for the number of bikes and check if 
#### the requested number is available or not if yes, function will deduct number_of_bikes 
#### from shop stock else it will print wait 10 mins. And you have to make sure that the age 
#### of the customer is over than 6 years. 
####  returnBike: Allows customers to return their bikes to the rental shop by taking 
#### rentalTime, num_of_bikes as parameters. And calculate the total bill cost by multiplying 
#### rentalTime * 40. Also, add the returned bikes to the stock store again 
####  addBike: adds bikes into the list 
####  totalCost: returns the total cost of bikes in the store 
#### class Bike that is composed of the following attributes: 
####  name 
####  price 
#### Here are some of the methods that you need to add to your class Bike; 
####  def __init__(self). 
####  Getter / setter for all attributes.

```python

class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_name(self):
        return self.name

    def get_age(self):
        return self.age

    def set_name(self, new_name):
        self.name = new_name

    def set_age(self, new_age):
        self.age = new_age


class Bike:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def get_name(self):
        return self.name

    def set_name(self, new_name):
        self.name = new_name

    def get_price(self):
        return self.price

    def set_price(self, new_price):
        self.price = new_price


class MainBikeRental:
    def __init__(self):
        self.bikes = []
        self.rentals = []

    def get_bikes(self):
        return self.bikes

    def set_bikes(self, bikes):
        self.bikes = bikes

    def get_rentals(self):
        return self.rentals

    def set_rentals(self, rentals):
        self.rentals = rentals

    def addBike(self, bike):
        self.bikes.append(bike)
        print(f"{bike.get_name()} added to the shop.")

    def requestBike(self, number_of_bikes, customer):
        if customer.get_age() <= 6:
            print("Customer must be over 6 years old.")
            return False

        if number_of_bikes <= 0:
            print("Please request at least one bike.")
            return False

        if number_of_bikes <= len(self.bikes):
            rented_bikes = self.bikes[:number_of_bikes]
            self.bikes = self.bikes[number_of_bikes:]

            self.rentals.append({
                "customer": customer.get_name(),
                "bikes": rented_bikes
            })

            print(f"{number_of_bikes} bike(s) rented successfully.")
            return True

        print("Wait 10 mins.")
        return False

    def returnBike(self, rentalTime, num_of_bikes):
        if not self.rentals:
            print("No bikes have been rented.")
            return

        rental = self.rentals[-1]
        rented_bikes = rental["bikes"]

        if num_of_bikes <= 0 or num_of_bikes > len(rented_bikes):
            print("Invalid number of bikes to return.")
            return

        returned_bikes = rented_bikes[:num_of_bikes]
        self.bikes.extend(returned_bikes)
        del rented_bikes[:num_of_bikes]

        total_bill = rentalTime * 40
        print(f"{num_of_bikes} bike(s) returned.")
        print(f"Total bill: {total_bill}")

        if len(rented_bikes) == 0:
            self.rentals.pop()

    def totalCost(self):
        total = 0

        for bike in self.bikes:
            total += bike.get_price()

        return total


customer = Customer("Menna", 22)

shop = MainBikeRental()

shop.addBike(Bike("Mountain Bike", 5000))
shop.addBike(Bike("Road Bike", 7000))
shop.addBike(Bike("City Bike", 4000))

print("Total stock value:", shop.totalCost())

shop.requestBike(2, customer)

print("Available bikes:", len(shop.get_bikes()))

shop.returnBike(3, 2)

print("Available bikes:", len(shop.get_bikes()))
print("Total stock value:", shop.totalCost())
```
